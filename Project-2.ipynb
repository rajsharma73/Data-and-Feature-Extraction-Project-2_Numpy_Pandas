# preprocessing the data to check null values 

```python
# shape 
df.shape  # (194, 64)

#info 
df.info()
```

```python
df.describe()
```

```python
# Which Country has the highest population :
population=df[df['population']==df['population'].max()]
population
```

```python
# what is the capital of country with highest poplation 
df[df['population']==df['population'].max()]['capital_city']
```

```python
# Which Country has the least population 
df[df['population']==df['population'].min()]['country']
```

```python
# Which is the Capital of Country with least population 
df[df['population']==df['population'].min()]['capital_city']
```

```python
# Give me top  5 Countries with highest democracy_score 
df.sort_values(by='democracy_score',ascending=False,inplace=True)
df['country'].head()
```

```python
#how many total regions are there :
df['region'].value_counts().count()
```

```python
# how many countries lies in eastern europe region: 
df['region'].value_counts()['Eastern Europe']

df[df['region']=='Eastern Europe']['country']
```

```python
#Who is the political leader of 5 most populated country
df.sort_values(by='population',ascending=False,inplace=True)
df['country'].head()
```

```python
#Who is the political leader of second most populated country
df[df['population']==df['population'].nlargest(2).iloc[1]]
```

```python
#how many Countries are there whose  political leaders are unknown
df[df['political_leader'].isna()]['country']
```

```python
# how many countries are there with REpublic in there full name 
count=0
def country_long(txt):
    global count
    if 'republic' in txt.lower():
        count+=1
    return txt
df['country_long'].apply(country_long)
```
